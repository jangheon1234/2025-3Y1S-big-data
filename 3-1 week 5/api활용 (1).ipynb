```python
import requests

url = 'http://openapi.seoul.go.kr:8088/(인증키)/json/SebcCollegeInfoKor/1/100/'

response = requests.get(url)
print(response.content)
```

```python
data = response.json()
```

```python
data
```

```python
import pandas as pd
temp = []
for d in data['SebcCollegeInfoKor']['row']:
    temp.append(d)
uni = pd.DataFrame(temp)
uni
```

```python
import requests

url = 'http://openapi.seoul.go.kr:8088/(인증키)/json/viewNightSpot/1/100/'

response = requests.get(url)
print(response.content)
```

```python
import folium

map = folium.Map(location=[37.450086, 127.167980],
                 zoom_start=11,
                 tiles='cartodbpositron')


for index, row in uni.iterrows():
    title = row['TITLE']
    lat = row['LA']
    lon = row['LO']


    folium.Marker(
        location=[lat, lon],
        popup=title,
        tooltip=title
    ).add_to(map)

map
```
